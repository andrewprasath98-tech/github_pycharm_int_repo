```python
def find_max(numbers):
    if not numbers:
        return None

    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest



print(find_max([12, 45, 2, 99, 23]))
```

```python
def remove_duplicates(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result



print(
    remove_duplicates(["apple", "banana", "apple", "orange", "banana"]))
```

```python
def two_sum(nums, target):
    seen = {}  # Store value and its index
    for index, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], index]
        seen[num] = index
    return []


# Test
print(two_sum([2, 7, 11, 15], 9))  # Output: [0, 1] (because 2 + 7 = 9)
```

```python
def char_frequency(s):
    freq = {}
    for char in s:
        if char == " ":
            continue
        freq[char] = freq.get(char, 0) + 1
    return freq



print(char_frequency("hello world"))
```

```python
def count_vowels(word):
    count = 0

    for letter in word:
        if letter in "aeiouAEIOU":
            count += 1

    return count


word = input("Enter the word: ")

result = count_vowels(word)

print("Count of vowels:", result)

```

```python
def count_word(sentence):
    words = sentence.split()
    count = len(words)
    return count


sentence = input("Enter a sentence: ")

result = count_word(sentence)

print("Total number of words:", result)


```

```python
n = int(input("Enter a number: "))

even_sum = 0
odd_sum = 0

for i in range(1, n + 1):

    if i % 2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum + i

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```
