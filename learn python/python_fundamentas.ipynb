## **D_I,D_TYPE,S_Typing**

```python
fee = 45000

# Dynamic inference: Python automatically infers the type of fee
print("Fee:", fee)
print("Type of fee:", type(fee))

# Calculate 18% GST
gst = fee * 0.18

print("GST (18%):", gst)
print("Total Fee:", fee + gst)

# Dynamic typing: the same variable can hold a different type
fee = "45000"
print("Fee as string:", fee)
print("Type of fee:", type(fee))

# Strong typing: Python does not automatically convert string to number
# The following will give a TypeError:
# gst = fee * 0.18

# Explicit type conversion makes the operation possible
fee = int(fee)
gst = fee * 0.18

print("GST after conversion:", gst)
print("Total Fee after conversion:", fee + gst)
```
