# **Use Case 1: Internet Data Usage Calculator**


```python

total_limit = float(input("Enter your total monthly data limit (GB): "))
data_used = float(input("Enter data used so far (GB): "))
Remaining_data = total_limit - data_used
Usage_percentage = (data_used / total_limit) * 100
print("Remaining data:",Remaining_data,"GB")
print("Usage percentage:", round(Usage_percentage, 2), "%")
if Usage_percentage>= 80:
   print("Warning: High usage,consider upgrading your plan.")

```

# **Use Case 2: Shopping Discount Calculation**


```python
Original_price=float(input("Enter Original_price:"))
Discount_percent=int(input("Enter discount_percentage:"))
Discount_amount =Original_price * (Discount_percent/ 100)
Final_price = Original_price - Discount_amount
print("Original price:", Original_price)
print("Discount amount:", Discount_amount)
print("Final price:", Final_price)
```
